```python
user=int(input("추측값을 입력하시오 "))
```

```python
count=0
while answer!=user:
  if answer>user:
    print("더 큰 수를 입력하시오")
    count+=1
  else:
    print("더 작은 수를 입력하시오")
    count+=1
  user=int(input("추측값을 입력하시오"))
print(count,"번만에 맞췄습니다.")
```
