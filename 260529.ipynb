```python
sepal_length_mean = df['SepalLengthCm'].mean()
print(f"The average sepal length is: {sepal_length_mean:.2f} cm")
```

```python
max_petal_width = df['PetalWidthCm'].max()
print(f"꽃잎 너비의 최댓값: {max_petal_width:.2f}")
```

```python
petal_length_greater_than_2 = df[df['PetalLengthCm'] > 2.0]
count = len(petal_length_greater_than_2)
print(f"꽃잎 길이가 2.0보다 큰 데이터의 개수: {count}개")
```

```python
버시컬러 = df[df['Species'] == 'Iris-versicolor']
평균꽃잎길이 = 버시컬러['PetalLengthCm'].mean()
print(f"Iris-versicolor의 평균 꽃잎 길이: {평균꽃잎길이:.2f} cm")
```

```python
버지니카 = df[df['Species'] == 'Iris-virginica']
최대꽃받침너비 = 버지니카['SepalWidthCm'].max()
print(f"Iris-virginica의 최대 꽃받침 너비: {최대꽃받침너비:.2f} cm")
```

```python
종류별평균 = df.groupby('Species')[['SepalLengthCm', 'PetalWidthCm']].mean()
display(종류별평균)
```

```python
df['꽃받침면적'] = df['SepalLengthCm'] * df['SepalWidthCm']
df['꽃잎면적'] = df['PetalLengthCm'] * df['PetalWidthCm']

면적평균 = df.groupby('Species')[['꽃받침면적', '꽃잎면적']].mean()
display(면적평균)
```

```python
필터링 = df[(df['SepalWidthCm'] >= 3.0) & (df['PetalWidthCm'] >= 1.5)]
display(필터링.head())
```

```python
종류별_분포 = 필터링['Species'].value_counts()
print("필터링된 데이터의 종류별 분포:")
print(종류별_분포)
```

```python
최대종 = 종류별_분포.idxmax()
최대종개수 = 종류별_분포.max()
최소종 = 종류별_분포.idxmin()
최소종개수 = 종류별_분포.min()

print(f"필터링된 데이터에서 가장 많이 포함된 종류: {최대종} ({최대종개수} 개)")
print(f"필터링된 데이터에서 가장 적게 포함된 종류: {최소종} ({최소종개수} 개)")
```
