```python
import random

dice1=random.randint(1,6)
dice2=random.randint(1,6)

print(f'주사위1의 값={dice1}')
print(f'주사위2의 값={dice2}')

if dice1>dice2:
  print("주사위1의 값이 더 크다.")
elif dice1==dice2:
  print("주사위들의 값이 같다.")
elif dice1<dice2:
  print("주사위2의 값이 더 크다.")
```

```python
#가위바위보(1=가위, 2=바위, 3=보)
import random

man1=random.randint(1,3)
man2=random.randint(1,3)

print(f'사람1:{man1}')
print(f'사람2:{man2}')

if man1==man2:
  print("무승부")
elif man1==1 and man2==3 or man1==2 and man2==1 or man1==3 and man2==2:
  print("사람1이 이김")
else:
  print("사람2가 이김")
```
